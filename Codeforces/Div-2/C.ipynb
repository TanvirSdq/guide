```python
from math import gcd

for _ in range(int(input())):
    n, x = map(int, input().split())
    arr = list(map(int, input().split()))

    divs = set()
    for d in range(1, int(x**0.5) + 1):
        if x % d == 0:
            if d > 1:
                divs.add(d)
            if x // d > 1:
                divs.add(x // d)

    ans = 0
    for d in divs:
        total = sum(a for a in arr if a % d == 0)
        if total > ans:
            ans = total

    print(ans)
```
