# Function

```python
#No parameter no return
def greet():
    print("Hello, World!")
greet()
```

```python
#With parameter no return
def greet(name = "None"):
    print(f"Hello, {name}!")
greet("Alice")
greet()
```

```python
def adder(x, y):
    """ArithmeticError: This function performs addition of two numbers."""
    return x + y

print(adder(5, 3))

def add(x, y):
    sum = x + y
    print(sum)
add(5, 3)


try:
    m = add(10,20) * 5
    print(m) 
except TypeError as e:
    print(f"Error: {e}.")

#but
m = adder(10,20) * 5
print(m)
```

## Arguments and Parameters

```python
# Positional arguments
def add(a, b, c):
    print(f"{a}, {b}, {c}")

add(1, 4, 3)

#but
add(b=1, c=4, a=3)
```

```python
# Many args
def add(*args):
    print(args)
    for i in args: # Two ways!
        print(i)

add(1, 2, 3, 4, 5)
```

```python
def add(**kwargs):
    print(kwargs)

    for key, value in kwargs.items():
        print(f"{key} = {value}")

add(a=1, b=2, d=4, n_estimator=50)
```

```python
# Some interesting examples
def adder(a,b):
    return a, b

print(type(adder(1,2)))

def adder(a,b):
    return a + b, a - b
print(type(adder(1,2)), adder(1,2))

def adder(a,b):
    return a, b

#unzip

x,y = adder(10.5 ,20)
print(x, y)
```

## Iterators and Generators

```python
some_list = [1, 2, 3, 4, 5]
print(some_list[1])

s = {10, 20, 30, 40, 50}
#print(s[1]) 

s_iter = iter(s)
print(s_iter)
print(type(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
# Random : This happens cuz set is not ordered.
for i in s:
    print(i) #This seems easy.
```

```python
list = [x for x in range (500)]

def data_loader(batch_size, list):
    for i in range(0, len(list), batch_size):
        yield list[i:i+batch_size]

x = data_loader(10, list)
print(type(x))
print(x)
print(next(x))
print(next(x))
```

## Lambda, Map, Apply Funcunctions

```python
# Lambda a : expression
import pandas as pd

df = pd.DataFrame({'Temperature (C)': [30, 32, 28.0, 25, 27]})
df['Temperature (F)'] = df['Temperature (C)'].apply(lambda x: (x * 9/5) + 32)
df
```

```python
df = pd.DataFrame({'A': [1, 2, 3, 4, 5], 'B': [10, 20, 30, 40, 50]})
df

df['B'] = df['B'].map(lambda x: x ** .5)
df
```
