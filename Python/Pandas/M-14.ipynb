```python
# Common Regex Filters for Pandas

# 1. Basic Character Matching
r"^[A-Z]"          # Starts with an uppercase letter
r"^[a-z]"          # Starts with a lowercase letter
r"^[A-Za-z]{4}$"   # Exactly 4 letters long
r"\d"              # Contains any digit (0-9)
r"\s"              # Contains a space

# 2. Advanced & Conditional Matching
r"(New|Los)"       # Contains "New" OR "Los"
r"[aeiou]{2}"      # Contains two consecutive vowels
r"^[^A]"           # Does NOT start with the letter "A"
r"^[A-Z][a-z]*$"   # A single capitalized word (no spaces)
r"\b\w{5}\b"       # Contains an exact 5-letter word

# 3. Custom Lengths & Patterns
r"^S.*a$"          # Starts with "S" and ends with "a"
```

### Apply Function

```python
# Min, Max Scaling

mn = df['Total Marks'].min()
mx = df['Total Marks'].max()

df['ScaledMarks'] = df['Total Marks'].apply(lambda x: (x - mn) / (mx - mn))
df
```

```python
# Custom Built Function

def grading_system(marks):
    if marks >= 260:
        return 'A+'
    elif marks >= 230:
        return 'A'
    elif marks >= 200:
        return 'B'
    elif marks >= 180:
        return 'C'
    else:
        return 'F'


df['Grade'] = df['Total Marks'].apply(grading_system)
df
```

```python
def grading_system(df):
    a = df['Math'] * 2
    b = df['Science'] * 3
    c = df['English'] * 4

    return a + b + c

df['Excellence'] = df.apply(grading_system, axis=1)
df
```
