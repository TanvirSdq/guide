```python
import pandas as pd

data = {
    "Name": ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry', 'Sophia', 'Jack'],
    "City": ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio', 'San Diego', 'H Los Angeles', 'Seol San Francisco'],
    "Department": ['HR', 'Finance', 'IT', 'Marketing', 'Sales', 'Operations', 'Legal', 'R&D', 'Finance', 'IT'],
    "Salary": [70000, 80000, 90000, 75000, 85000, 95000, 72000, 88000, 91000, 78000]
}

df = pd.DataFrame(data)
df
```

```python
df.loc[df['City'].str.contains("San", case = False)] #Case Sensitive if not specified
```

```python
df.loc[df['City'].str.contains("san")]
```

```python
# Using Regex
df.loc[df['City'].str.contains(r"^Los")] # ^ Just starts with
```

```python
df.loc[df['City'].str.contains(r"es|go$")] # Last letters
```

```python
df.loc[df['Name'].str.contains(r"^[AEIOU]")] # Names starting with vowels
```

```python
df.loc[df['City'].str.contains(r'New|Los|San')] # All that contains New, Los or San
```

```python
# Common Regex Filters for Pandas

# 1. Basic Character Matching
r"^[A-Z]"          # Starts with an uppercase letter
r"^[a-z]"          # Starts with a lowercase letter
r"^[A-Za-z]{4}$"   # Exactly 4 letters long
r"\d"              # Contains any digit (0-9)
r"\s"              # Contains a space

# 2. Advanced & Conditional Matching
r"(New|Los)"       # Contains "New" OR "Los"
r"[aeiou]{2}"      # Contains two consecutive vowels
r"^[^A]"           # Does NOT start with the letter "A"
r"^[A-Z][a-z]*$"   # A single capitalized word (no spaces)
r"\b\w{5}\b"       # Contains an exact 5-letter word

# 3. Custom Lengths & Patterns
r"^S.*a$"          # Starts with "S" and ends with "a"
```
