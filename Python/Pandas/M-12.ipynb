### Transforming inbuilt data to a dataframe

```python
my_list = [['Alice', 24], ['Bob', 19], ['Charlie', 22]]
list_df = pd.DataFrame(my_list, columns=['Name', 'Age'], index=[1, 2, 3])
print(type(list_df))
list_df
```

```python
my_tuple = [('Alice', 24), ('Bob', 19), ('Charlie', 22)]
tuple_df = pd.DataFrame(my_tuple, columns=['Name', 'Age'], index=[1, 2, 3])
print(type(tuple_df))
tuple_df
```

```python
my_dict = [{'Name': 'Alice', 'Age': 24, 'City': 'New York'},
          {'Name': 'Bob', 'Age': 19, 'City': 'Los Angeles'},
          {'Name': 'Charlie', 'Age': 22}]
dict_df = pd.DataFrame(my_dict,index=[1, 2, 3]) # Missing Key doesn't matter and column name is autoguessed
print(type(dict_df))
dict_df # Json is basically a dictionary in python. 
```
