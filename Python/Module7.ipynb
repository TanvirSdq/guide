```python
# with open closes the file automatically
with open("samples/data.txt", "r") as file:
    for line in file:
        print(line.strip())
        #or 
        print(line, end="")
```

```python
with open("samples/data.txt", "w") as file:
    file.write("Hello World!")
    file.write("\nI am doing odd jobs in Python.")
# This will overwrite the existing content of the file.
with open("samples/data.txt", "a") as file:
    file.write("\nIt is a beautiful day.")
# This will append the content to the end of the file.
```

```python
content = list(map(str.strip, content))
print(content)
filter_content = filter(lambda x: len(x) >= 1, content)
print(list(filter_content))
```

```python
list_of_strings = ['\n','Hello', 'World', 'Python', 'is', 'awesome', 'Whatsup?']
with open("samples/data.txt", "a") as file:
    file.writelines([s + "\n" for s in list_of_strings])
```

```python
with open("samples/data.txt", "r") as file:
    print(file.tell())
    print(file.seek(0))
    print(file.read())
    print(file.tell())
    print(file.read()) #cursor is at the end of the file
```

```python
with open("samples/data.txt", "r") as file:
    print(file.seek(0))
    print(file.tell())
    print(file.read(10)) #read first 10 characters
    print(file.tell())
    print(file.readline()) #read the next line
```

```python
from functools import reduce
print(num_of_words)
total_number_of_words = reduce(lambda x, y: x + y, num_of_words)
print(total_number_of_words)

#new line deleted
strings = list(map(str.strip, string))
print(strings)

string_list = map(lambda x: x.replace(" ", ""), strings)
print(list(string_list))

total_char = 0
for line in strings:
    total_char += len(line)
print(total_char)
```
