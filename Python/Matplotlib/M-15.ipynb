```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## 2D line plot

```python
hours = [4, 6, 5, 3, 5, 7, 4]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study Hours Over a Week')
plt.grid(True)
plt.plot(days, hours, marker='o', linestyle='-', color='skyblue')
```

```python
hours_1 = [4, 6, 5, 3, 5, 7, 4]
days = [1, 2, 3, 4, 5, 6, 7]
hours_2 = [5, 7, 6, 4, 6, 8, 5]

plt.plot(days, hours_1, marker='o', linestyle='-', color='skyblue', label='Student 1')
plt.plot(days, hours_2, marker='s', linestyle='--', color='lightcoral', label='Student 2')
```

# Scatter Plot

```python
students = pd.read_csv('../samples/student_iq.csv')
students
```

```python
# Positive correlation between study hours and IQ score

plt.scatter(students['Study_Hour'], students['IQ_Score'])
plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.title('Study Hours vs IQ Score')
```

```python
# No correlation between shoe size and IQ score

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
plt.xlabel('Shoe Size')
plt.ylabel('IQ Score')
plt.title('Shoe Size vs IQ Score')
```

```python
# Negative correlation between chilling hours and IQ score

plt.scatter(students['Chilling_Hours'], students['IQ_Score'], color='orange', marker='x')
plt.xlabel('Chilling Hours')
plt.ylabel('IQ Score')
plt.title('Chilling Hours vs IQ Score')
```

## Histogram

```python
plt.hist(students['IQ_Score'], bins=15, color='lightblue', edgecolor='black')
plt.xlabel('IQ Score')
plt.ylabel('Frequency')
plt.title('Distribution of IQ Scores')
```

```python
### Most of the data has outliers

# Based on study hours

plt.boxplot(students['Study_Hour'], patch_artist=True, boxprops=dict(facecolor='lightblue', color='blue'), medianprops=dict(color='red'), whiskerprops=dict(color='blue'), capprops=dict(color='blue'))
plt.title('Boxplot of Study Hours')
plt.ylabel('Study Hours')
```

```python
plt.hist(students['Study_Hour'], bins=15, color='lightblue', edgecolor='black')
plt.title('Distribution of Study Hours')
plt.ylabel('Frequency')
plt.xlabel('Study Hours')

# bin size can be given as bins = [0, 2, 4, 6, 8, 10, 12]
```

```python
# bar chart
df = pd.read_csv('../samples/student_data.csv')
df
```

```python
# Genderwise

gender_group = df.groupby('Gender').size() # Size includes NaN, Count doesn't
gender_group
```

```python
plt.bar(gender_group.index, gender_group.values, color=['lightblue', 'lightcoral'], edgecolor='black')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.title('Genderwise Distribution of Students')
```

```python
dept_group = df.groupby('Department').size()
plt.bar(dept_group.index, dept_group.values, color='lightgreen', edgecolor='black')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.title('Departmentwise Distribution of Students')
```

```python
place = df.groupby('HomeTown').size()
plt.bar(place.index, place.values, color='lightyellow', edgecolor='black', width=1.0)
plt.xlabel('HomeTown')
plt.ylabel('Number of Students')
plt.title('Hometownwise Distribution of Students')
```

## Pie chart

```python
plt.pie(place.values, labels=place.index, autopct='%1.1f%%', startangle=90, colors=['lightblue', 'lightcoral', 'lightgreen', 'lightyellow', 'lightpink'])
```

```python
plt.pie(dept_group.values, labels=dept_group.index, autopct='%1.1f%%', startangle=90, colors=['lightblue', 'lightcoral', 'lightgreen', 'lightyellow', 'lightpink'])
plt.title('Departmentwise Distribution of Students')
plt.show()
```

```python
# Shadow, explode, startangle, colors can be used in pie chart
plt.pie(dept_group.values, labels=dept_group.index, autopct='%1.1f%%', startangle=90, shadow=True, explode=(0, 0, 0, 0.1, 0, 0))
plt.title('Departmentwise Distribution of Students')
plt.show()
```
