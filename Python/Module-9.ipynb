# OOP

### Class and Object

```python
class Phone:
    category = "Electronics" # Class attribute

    #Constructor
    def __init__(self, model, battery, camera, battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    def charge(self, hour):
        self.battery_percentage = min(100, self.battery_percentage + hour)
        print(f"Phone charged to {self.battery_percentage}%.")
        return self.battery_percentage

    def capture(self, photo):
        if self.battery_percentage <= 0:
            print("Battery is empty. Please charge your phone.")

        else:
            self.battery_percentage -= photo
            print(f"Photo captured in {self.model}.")
        
    
```

```python
# Creating Objects
apple = Phone("iPhone 17", "5000mAh", "108MP")
samsung = Phone("Galaxy S23", "4500mAh", "64MP")
print(type(apple))

print(apple.category) 
print(samsung.category)
print(apple.model)
print(apple.battery)
print(apple.camera)
apple.capture(10)
```

# Inheritencance

```python
class Smartphone(Phone):
    def __init__(self, model, battery, camera, battery_percentage=100, os="Android", processor="Snapdragon"):
        super().__init__(model, battery, camera, battery_percentage)
        self.os = os
        self.processor = processor



    def charge(self, hour):
        print("Fast Charging...")
        super().charge(hour)

    def install_app(self, app_name):
        if self.battery_percentage <= 0:
            print("Battery is empty. Please charge your phone.")
        else:
            self.battery_percentage -= 5
            print(f"{app_name} installed on {self.model}. Battery percentage is now {self.battery_percentage}%.")
```

```python
pro = Smartphone("X", "4000mAh", "48MP", os="iOS", processor="A15 Bionic")
print(pro.model)
pro.install_app("Instagram")
pro.charge(1)
```

## Multi Inheritritance

```python
class CoolingMechanism:
    def __init__(self, cooling_method):
        self.cooling_method = cooling_method

    #methods
    def cooling_on(self):
        print(f"Cooling mechanism activated using {self.cooling_method}.")
            
```

```python
class Smartphone_CoolingMode(Smartphone, CoolingMechanism):
    def __init__(self, model, battery, camera, battery_percentage=100, os="Android", processor="Snapdragon", cooling_method="Liquid Cooling"):
        Smartphone.__init__(self, model, battery, camera, battery_percentage, os, processor)
        CoolingMechanism.__init__(self, cooling_method)
```

```python
pro_cooling = Smartphone_CoolingMode("Y", "5000mAh", "108MP", os="iOS", processor="A15 Bionic", cooling_method="Liquid Cooling")
print(pro_cooling.model)
pro_cooling.charge(1)
print(pro_cooling.battery_percentage)
print(pro_cooling.processor) 

pro_cooling.cooling_on()
# This is how multi inheritance works in Py.
```

# Polymorphism

```python
class Camera:
    def __init__(self, name):
        self.name = name

    def capture(self):
        print(f"{self.name} camera is capturing an image.")
# We can have smartphone, dslr, drone cameras. Their capture form is varied.

class Smart(Camera):
    def __init__(self, name, resolution):
        super().__init__(name)
        self.resolution = resolution

    #method overriding
    def capture(self):
        print(f"{self.name} smartphone camera is capturing an image with {self.resolution} resolution.")

class Drone(Camera):
    def __init__(self, name, max_altitude):
        super().__init__(name)
        self.max_altitude = max_altitude

    #method overriding
    def capture(self):
        print(f"{self.name} drone camera is capturing an image at a maximum altitude of {self.max_altitude} meters.")

class DSLR(Camera):
    def __init__(self, name, lens_type):
        super().__init__(name)
        self.lens_type = lens_type

    #method overriding
    def capture(self):
        print(f"{self.name} DSLR camera is capturing an image with a {self.lens_type} lens.")

phone = Smart("Phone", 30)
drone = Drone("Drone", 500)
dslr = DSLR("DSLR", "Wide Angle")

phone.capture()  
drone.capture()
dslr.capture()
```

# Encapsulation

```python
class Mobile:
    def __init__(self, name, model, imei):
        self.__name = name
        self.__model = model
        self.__imei = imei #Private

    def charge(self):
        print(f"{self.name} is charging.")

    #Getter method to access the private attribute
    def get_imei(self):
        return self.__imei

    #Getter method to access the private attribute
    def get_model(self):
        return self.__model

    #Getter method to access the private attribute
    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

iPhone = Mobile("iPhone", "X", "123456789012345")
print(iPhone.get_model())
print(iPhone.get_name())
print(iPhone.get_imei())

iPhone.name = "Phone"
iPhone.model = "Y"
iPhone.imei = "543210987654321"

iPhone.set_name("iPhone 14")

print(iPhone.get_model())
print(iPhone.get_name())
print(iPhone.get_imei())  # Accessing the private attribute using a getter method
```

# Abstraction

```python
# Hiding  Complex Methods and Attributes
from abc import ABC, abstractmethod
class Telephone(ABC):
    @abstractmethod
    def make_call(self):
        pass
    @abstractmethod
    def capture_photo(self):
        pass
#This is the concept of abstraction. 
class Sphone(Telephone):
    def make_call(self):
        print("Making a call from Sphone.")

    def capture_photo(self):
        print("Capturing photo from Sphone.")

class Iphone(Telephone):
    def make_call(self):
        print("Making a call from Iphone.")

    def capture_photo(self):
        print("Capturing photo from Iphone.")


sphone = Sphone()
iphone = Iphone()
    
```
