```python
import numpy as np

arr = np.random.randint(1, 100, size=(10, 5)) # sample -> 10, features -> 5

arr.shape
```

```python
b = arr.reshape(5, 10) # Total elements should be equal
b.shape
b.ndim
b
```

```python
flatten = b.flatten() 
print(flatten)
print(flatten.ndim)
```

```python
columnwise_flatten = b.ravel(order='F') # Ravel can do both
print(columnwise_flatten)
print(columnwise_flatten.ndim)
```

```python
a = np.random.randint(1, 10, size=(2, 3))
b = np.random.randint(20, 30, size=(2, 3))

print(a, '\n\n', b)


# Concatination
c = np.concatenate((a, b), axis=0) # axis=0 -> row
print('\n', c)

d = np.concatenate((a, b), axis=1) # axis=1 -> column
print('\n', d) 
```

```python
# Errors
x = np.random.randint(1, 10, size=(2, 3))
y = np.random.randint(20, 30, size=(3, 2))

# rowwise concat needs same number of columns
# columnwise concat needs same number of rows

try:
    r = np.concatenate((x, y), axis=0) # axis=0 -> row
    c = np.concatenate((x, y), axis=1) # axis=1 -> column
    print('\n', r, '\n\n', c)
except ValueError as e:
    print(f"Error: {e}")
```

### Necessity of Transpose

##### We can add multiple features in a matrix

```python
mat = np.array([[10, 20, 30],
                 [40, 50, 60]
                 ])

transposed_mat = mat.T #or np.transpose(mat)
print(transposed_mat)
```

```python
# Array splitting
arr = np.random.randint(1, 10, size=(10,)) 
print(arr)
split_arr = np.array_split(arr, 3)
print(split_arr)

try:
    splitted_arr = np.split(arr, 3) # Error if not divisible
    print(splitted_arr)
except ValueError as e:
    print(f"Error: {e}")
```

## NumPy Arithmetics

```python
# Faster for vectorized operations

x = np.array([10, 20, 30, 40, 50, 60])
y = np.array([7, 14, 17, 23, 26, 31])

# There is another numpy add for arrays -> np.add(x, y) 
add = x + y
print(add)

# Another way to do this is using -> np.subtract(x, y)
sub = x - y
print(sub)

# Another way to do this is using -> np.multiply(x, y)
mul = x * y
print(mul)

# Another way to do this is using -> np.divide(x, y)
div = x / y
print(div)

# Another way to do this is using -> np.mod(x, y)
rem = x % y
print(rem)
```

```python
# Mathematical functions

# Trigonometric functions
sin_val = np.sin(x)
print(sin_val) # Values in radians

tan_val = np.tan(x)
print(tan_val) # Values in radians

cos_val = np.cos(x)
print(cos_val) # Values in radians

# Rad to Degrees
rad_to_deg = np.degrees(x)
print(rad_to_deg)
#or
rad_to_deg = np.rad2deg(x)
print(rad_to_deg)
```

```python
# Log
x = np.array([1, 10, 100, 1000, 10000])
base_10_log = np.log10(x)
print(base_10_log)

base_2_log = np.log2(x)
print(base_2_log)

sqrt_val = np.sqrt(x)
print(sqrt_val)

sum_val = np.sum(x)
print(sum_val)

cum_sum = np.cumsum(x) #or np.cumulative_sum(x)
print(cum_sum)
```

### Logical functions

```python
x = np.array([10, 13, 17, 22, 24])
y = np.array([6,  14, 18, 21, 25])

greater_than = x > y
print(greater_than)

equal_to = x == y
print(equal_to)

# all() -> for all true values
# any() -> for any true value

print(np.all(greater_than))
print(np.any(greater_than))

# Other ways

print(np.greater(x, y))
print(np.equal(x, y))
```

## Sorting

```python
#inplace
x = np.array([10, 8, 16, 13, 19, 30, 24])
z = x.copy()
print(z)

z.sort() #inplace sorting
print(z)
```

```python
# Copy sort
print(x) 
sort_x = np.sort(x) #copy sort
print(sort_x)
print(x) 
```

```python
# 2D array sorting
mat = np.array([[13, 7, 9],
                 [23, 11, 5]
                 ])

print("Original matrix:")
print(mat)

#horizon_axis = 1
hor_sort = np.sort(mat, axis=1) 
print("Matrix sorted along horizontal axis:")
print(hor_sort)

#vertical_axis = 0
print("Matrix sorted along vertical axis:")
ver_sort = np.sort(mat, axis=0)
print(ver_sort)
```
