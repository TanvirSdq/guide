## Indexing and Slicing

#### String[Start : End : Step]

```python
first_message = message[0:10]
print(first_message)
```

```python
#Negative Indexing
print(first_message[-1])
print(first_message[-2])
print(first_message[-3])
```

```python
new_message = first_message[::-1]
print(new_message)
```

## String Methods

```python
string = "Why the hell am I in Hell?"
print(len(string))
process_string = string.lower()

print("hell" in process_string)
process_string
```

```python
processed_string = process_string.capitalize()
print(processed_string)

#Knowing which index the word is in the string
index = process_string.find("hell")
print(index)
rindex = process_string.rfind("hell")
print(rindex)

print(process_string.count("hell"))
```
